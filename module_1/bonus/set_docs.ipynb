# Множини (Sets) у Python: Детальна документація та приклади

**Множина (set)** у Python — це неупорядкована колекція унікальних об'єктів. Вона працює так само, як математичні множини, і підтримує відповідні операції (перетин, об'єднання тощо).

Множини можна уявити як мішок, у який ви закидаєте речі: 
1. Ви не знаєте, в якому порядку вони там лежать.
2. Якщо ви покладете туди два однакових яблука, "мішок" залишить лише одне. Усі елементи абсолютно унікальні.

## 1. Базові принципи та створення
* **Швидкість:** Перевірка, чи є елемент у множині (`x in my_set`), працює миттєво ($O(1)$) завдяки хеш-таблицям під капотом.
* **Елементи:** Повинні бути незмінними (можна додати числа або рядки, але не можна додати список або словник).

```python
# 1. Створення множини
fruits = {"яблуко", "банан", "апельсин"}
print("Початкова множина:", fruits)

# 2. ⚠️ ВАЖЛИВИЙ НЮАНС для новачків: Порожня множина
# Щоб створити порожню множину, треба використовувати set(), а не фігурні дужки!
empty_dict = {}     # Це створить порожній словник
empty_set = set()   # Це створить порожню множину

print("\nТип empty_dict:", type(empty_dict))
print("Тип empty_set:", type(empty_set))

# 3. Швидке видалення дублікатів зі списку
dirty_data = [1, 2, 2, 3, 3, 3, 4, 5, 5]
clean_data = list(set(dirty_data))
print("\nОчищений список від дублікатів:", clean_data)
```

## 2. Додавання та видалення елементів

Множини (set) є змінюваними об'єктами, тому ви можете керувати їхнім вмістом після створення.

* `add(x)` — додає один елемент `x`. Якщо він вже існує, метод нічого не робить.
* `remove(x)` — видаляє елемент `x`. **Увага:** якщо елемента немає, викличе помилку `KeyError`.
* `discard(x)` — "безпечно" видаляє елемент `x`. Якщо його не існує, метод просто ігнорує команду без помилок.
* `pop()` — видаляє та повертає один *довільний* елемент. Якщо множина порожня, викликає `KeyError`.
* `clear()` — повністю очищає множину, залишаючи `set()`.

```python
# Управління активними сесіями користувачів
active_users = {"admin", "moderator", "guest"}

# Додавання
active_users.add("new_user")
active_users.add("admin") # Нічого не зміниться, дублікат ігнорується
print("Після додавання:", active_users)

# Видалення
active_users.remove("guest")

# Безпечне видалення (користувача 'hacker' немає, але discard не зламає програму)
active_users.discard("hacker") 
print("Після видалення:", active_users)

# Вилучення довільного елемента
random_user = active_users.pop()
print(f"Вилучено користувача: {random_user}")
print("Залишилось:", active_users)

# Очищення
active_users.clear()
print("Після clear():", active_users)
```

## 3. Математичні операції (Створення нових множин)

Ці методи реалізують класичну теорію множин. Вони не змінюють оригінальні множини, а створюють нову. 
**Важливий нюанс:** Методи (наприклад, `.union()`) можуть приймати будь-який ітерований об'єкт (список, рядок). Оператори (наприклад, `|`) вимагають, щоб з обох боків були саме множини.

### Об'єднання (Union) `union(other)` або `|`

Повертає нову множину, що містить всі унікальні елементи з обох множин.

### Перетин (Intersection) `intersection(other)` або `&`

Повертає нову множину лише з тих елементів, які є одночасно в обох множинах.

### Різниця (Difference) `difference(other)` або `-`


[Image of Venn diagram showing set difference]

Повертає елементи першої множини, яких немає у другій.

### Симетрична різниця (Symmetric Difference) `symmetric_difference(other)` або `^`

Повертає елементи, які є в будь-якій з множин, але не в обох одночасно.

```python
# --- ПРИКЛАД: Соціальна мережа та Завдання ---
set_A = {"Антон", "Максим", "Юлія"}
set_B = {"Максим", "Світлана"}

print("Об'єднання (|):", set_A | set_B) # Всі унікальні люди
print("Перетин (&):", set_A & set_B)    # Спільні (тільки Максим)
print("Різниця (A - B):", set_A - set_B) # Є в A, але немає в B (Антон, Юлія)
print("Сим. різниця (^):", set_A ^ set_B) # Всі, окрім Максима

# Демонстрація гнучкості методів (передаємо список у метод union)
new_people_list = ["Олег", "Марія", "Антон"]
expanded_set = set_A.union(new_people_list) 
print("\nМетод union() зі списком:", expanded_set)
```

## 4. Оновлення множини "на місці" (Мутуючі операції)

Якщо множини дуже великі, створювати нові об'єкти в пам'яті неефективно. Ці методи записують результат прямо в поточну множину:

* `update(other)` або `|=` — об'єднання на місці (додає всі елементи з `other`).
* `intersection_update(other)` або `&=` — залишає лише спільні елементи.
* `difference_update(other)` або `-=` — видаляє елементи, присутні в `other`.
* `symmetric_difference_update(other)` або `^=` — записує симетричну різницю.

```python
# Збираємо унікальні IP-адреси з логів
seen_ips = {"192.168.1.1", "10.0.0.5"}
new_logs_batch = ["10.0.0.5", "172.16.0.2", "192.168.1.1"]

# update() ефективно додасть нові IP без створення нової множини в пам'яті
seen_ips.update(new_logs_batch)
print("Оновлені IP-адреси:", seen_ips)

# difference_update() - видаляємо внутрішні IP, залишаємо лише зовнішні
internal_ips = {"192.168.1.1", "10.0.0.5"}
seen_ips.difference_update(internal_ips)
print("Тільки зовнішні IP:", seen_ips)
```

## 5. Логічні перевірки (Відношення між множинами)

Ці методи завжди повертають `True` або `False` і дуже корисні для валідації даних:

* `issubset(other)` або `<=` — чи всі елементи першої множини містяться у другій (підмножина).
* `issuperset(other)` або `>=` — чи містить перша множина всі елементи іншої (надмножина).
* `isdisjoint(other)` — чи правда, що множини взагалі **не мають** спільних елементів.

```python
# ПРИКЛАД: Валідація колонок у таблиці даних
required_columns = {"id", "timestamp", "value"}
provided_columns = {"id", "timestamp", "value", "sensor_name", "status"}

# Перевіряємо, чи всі обов'язкові колонки присутні у наданих
is_valid = required_columns.issubset(provided_columns)
print(f"Чи валідна структура даних? {is_valid}")

# Перевірка на конфлікт інтересів (isdisjoint)
# Якщо перетин порожній - поверне True (спільних проектів немає)
frontend_tasks = {"ui_button", "css_fix"}
backend_tasks = {"api_endpoint", "db_migration"}
no_overlap = frontend_tasks.isdisjoint(backend_tasks)
print(f"Чи незалежні команди? {no_overlap}")
```

## 6. Інші корисні операції (`len`, `copy`, `frozenset`)

* `len(s)` — вбудована функція, повертає кількість елементів.
* `copy()` — створює поверхневу копію (shallow copy) множини.
* `frozenset()` — створює незмінну версію множини. Звичайну множину `set` не можна використовувати як ключ у словнику (бо вона змінювана). `frozenset` вирішує цю проблему, оскільки є статичною та хешованою.

```python
my_set = {10, 20, 30}
print(f"Розмір множини: {len(my_set)} елементів")

set_copy = my_set.copy()

# Використання frozenset як ключа у словнику кешування
# Порядок міст у frozenset не має значення! Ми знайдемо маршрут незалежно від порядку.
routes_cache = {
    frozenset(["Київ", "Чернівці"]): 540,
    frozenset(["Львів", "Чернівці"]): 270
}

search_key = frozenset(["Чернівці", "Київ"])
print(f"\nВідстань: {routes_cache[search_key]} км")
```

## 7. Просунуті приклади використання (Для Інженерії Даних та ML)

Реальні сценарії, де множини демонструють свою максимальну ефективність: оптимізація ETL-пайплайнів та підготовка даних для навчання нейромереж (SFT, DPO).

```python
import time

# --- ПРИКЛАД 1: Оптимізація швидкості пошуку в ETL-пайплайні ---
# Збираємо гідрологічні дані з датчиків для проекту GeoHydroAI.
processed_ids_list = list(range(10_000_000)) # 10 млн записів
processed_ids_set = set(processed_ids_list)
new_sensor_id = 9_999_999

# Пошук у списку O(N)
start_time = time.time()
_ = new_sensor_id in processed_ids_list
list_time = time.time() - start_time

# Пошук у множині O(1)
start_time = time.time()
_ = new_sensor_id in processed_ids_set
set_time = time.time() - start_time

print(f"Час пошуку в List: {list_time:.5f} сек")
print(f"Час пошуку в Set:  {set_time:.5f} сек")
print(f"Множина швидша приблизно у {list_time / (set_time + 1e-9):,.0f} разів!\n")

# --- ПРИКЛАД 2: Валідація токенів під час підготовки LLM ---
corpus_base = "штучний інтелект аналізує гідрологічні дані"
corpus_new = "гідрологічні дані важливі для нейромереж"

vocab_base = set(corpus_base.split())
vocab_new = set(corpus_new.split())

# Які нові терміни з'явилися у новому датасеті? (Різниця)
new_unique_tokens = vocab_new - vocab_base
print(f"Нові токени для донавчання: {new_unique_tokens}")
```

## 8. Повний аналіз математичних операцій над множинами

У розділі 3 були чотири основні операції. Тут — **повна картина**: усі операції теорії множин, як кожна записується в Python, які має властивості й де на ній помиляються.

Позначення: $A$ і $B$ — множини, $\varnothing$ — порожня множина, $U$ — **універсум** (усі елементи, про які взагалі йдеться в задачі).

| Операція | Математика | Оператор | Метод | Версія «на місці» | Результат |
|---|---|---|---|---|---|
| Об'єднання | $A \cup B$ | `A \| B` | `A.union(B, ...)` | `A \|= B`, `A.update(B)` | множина |
| Перетин | $A \cap B$ | `A & B` | `A.intersection(B, ...)` | `A &= B`, `A.intersection_update(B)` | множина |
| Різниця | $A \setminus B$ | `A - B` | `A.difference(B, ...)` | `A -= B`, `A.difference_update(B)` | множина |
| Симетрична різниця | $A \,\triangle\, B$ | `A ^ B` | `A.symmetric_difference(B)` | `A ^= B`, `A.symmetric_difference_update(B)` | множина |
| Доповнення | $\overline{A} = U \setminus A$ | `U - A` | `U.difference(A)` | — | множина |
| Належність | $x \in A$ | `x in A` | — | — | `bool` |
| Рівність | $A = B$ | `A == B` | — | — | `bool` |
| Підмножина | $A \subseteq B$ | `A <= B` | `A.issubset(B)` | — | `bool` |
| Власна підмножина | $A \subset B$ | `A < B` | — | — | `bool` |
| Надмножина | $A \supseteq B$ | `A >= B` | `A.issuperset(B)` | — | `bool` |
| Власна надмножина | $A \supset B$ | `A > B` | — | — | `bool` |
| Неперетинні | $A \cap B = \varnothing$ | — | `A.isdisjoint(B)` | — | `bool` |
| Потужність | $\lvert A \rvert$ | `len(A)` | — | — | `int` |
| Декартів добуток | $A \times B$ | — | `itertools.product(A, B)` | — | пари |
| Булеан (усі підмножини) | $\mathcal{P}(A)$ | — | через `itertools.combinations` | — | множина множин |

Нижче кожна операція розібрана окремо. Комірка одразу під таблицею показує всі чотири основні операції на одній парі множин — щоб було з чим порівнювати далі.

> У прикладах множини друкуються через `sorted(...)`: порядок елементів у множині не гарантований, а відсортований список щоразу виглядає однаково.

```python
python_course = {"Оля", "Марко", "Іра", "Денис"}
sql_course = {"Іра", "Денис", "Софія"}

print("A (Python):", sorted(python_course))
print("B (SQL):   ", sorted(sql_course))
print()
print("A | B  об'єднання:        ", sorted(python_course | sql_course))
print("A & B  перетин:           ", sorted(python_course & sql_course))
print("A - B  різниця:           ", sorted(python_course - sql_course))
print("B - A  різниця навпаки:   ", sorted(sql_course - python_course))
print("A ^ B  симетрична різниця:", sorted(python_course ^ sql_course))
```

### 8.1. Об'єднання $A \cup B$ — «хоча б в одній»

$$A \cup B = \{x \mid x \in A \ \text{або} \ x \in B\}$$

Результат містить кожен елемент, який є **хоча б в одній** із множин. Спільні елементи потрапляють у результат один раз.

**Запитання, на яке відповідає:** «хто взагалі є?», «усі унікальні значення з кількох джерел».

**Властивості:**

| Властивість | Запис | Що означає на практиці |
|---|---|---|
| Комутативність | $A \cup B = B \cup A$ | порядок множин не важливий |
| Асоціативність | $(A \cup B) \cup C = A \cup (B \cup C)$ | об'єднувати можна в будь-якому порядку і по кілька одразу |
| Ідемпотентність | $A \cup A = A$ | об'єднання із собою нічого не змінює |
| Нейтральний елемент | $A \cup \varnothing = A$ | порожня множина — зручний початок для накопичення |
| Розмір | $\max(\lvert A\rvert, \lvert B\rvert) \le \lvert A \cup B\rvert \le \lvert A\rvert + \lvert B\rvert$ | результат не менший за більшу з множин |

**У Python:** `A | B` або `A.union(B, C, ...)` — метод приймає одразу кілька аргументів і не тільки множини.

```python
python_course = {"Оля", "Марко", "Іра", "Денис"}
sql_course = {"Іра", "Денис", "Софія"}
git_course = {"Софія", "Тарас"}

everyone = python_course | sql_course | git_course
print("Усі студенти школи:", sorted(everyone))
print("Те саме методом:   ", sorted(python_course.union(sql_course, git_course)))

print()
print("Розміри: 4 + 3 + 2 =", len(python_course) + len(sql_course) + len(git_course),
      "записів, але унікальних людей", len(everyone))
```

```python
A, B, C = {1, 2, 3}, {3, 4}, {4, 5}

print("Комутативність:   A | B == B | A            →", A | B == B | A)
print("Асоціативність:   (A | B) | C == A | (B | C) →", (A | B) | C == A | (B | C))
print("Ідемпотентність:  A | A == A                 →", A | A == A)
print("Нейтральний:      A | set() == A             →", A | set() == A)

# Накопичення: починаємо з порожньої множини і «доливаємо» кожну нову порцію
seen = set()
for batch in ([1, 2], [2, 3], [3, 3, 4]):
    seen |= set(batch)
    print(f"після порції {batch}: {sorted(seen)}")
```

### 8.2. Перетин $A \cap B$ — «в обох одночасно»

$$A \cap B = \{x \mid x \in A \ \text{і} \ x \in B\}$$

Результат містить лише ті елементи, які є **в кожній** із множин.

**Запитання, на яке відповідає:** «що спільного?», «хто є і там, і там?».

**Властивості:**

| Властивість | Запис | Що означає на практиці |
|---|---|---|
| Комутативність | $A \cap B = B \cap A$ | порядок не важливий |
| Асоціативність | $(A \cap B) \cap C = A \cap (B \cap C)$ | можна перетинати кілька множин одразу |
| Ідемпотентність | $A \cap A = A$ | перетин із собою нічого не змінює |
| Поглинальний елемент | $A \cap \varnothing = \varnothing$ | одна порожня множина «обнуляє» весь результат |
| Розмір | $0 \le \lvert A \cap B\rvert \le \min(\lvert A\rvert, \lvert B\rvert)$ | результат не більший за меншу з множин |

Якщо перетин порожній, множини називають **неперетинними** (*disjoint*). Для цієї перевірки є окремий метод `A.isdisjoint(B)`: він зупиняється на першому ж спільному елементі й не будує проміжну множину.

**У Python:** `A & B` або `A.intersection(B, C, ...)`.

```python
python_course = {"Оля", "Марко", "Іра", "Денис"}
sql_course = {"Іра", "Денис", "Софія"}
git_course = {"Софія", "Тарас", "Іра"}
design_course = {"Юля", "Роман"}

print("І Python, і SQL:      ", sorted(python_course & sql_course))
print("Усі три курси одразу: ", sorted(python_course & sql_course & git_course))
print("Те саме методом:      ", sorted(python_course.intersection(sql_course, git_course)))

print()
print("Python і дизайн не перетинаються:", python_course.isdisjoint(design_course))
print("Python і SQL не перетинаються:   ", python_course.isdisjoint(sql_course))
```

```python
A, B, C = {1, 2, 3}, {2, 3, 4}, {3, 4, 5}

print("Комутативність:   A & B == B & A            →", A & B == B & A)
print("Асоціативність:   (A & B) & C == A & (B & C) →", (A & B) & C == A & (B & C))
print("Ідемпотентність:  A & A == A                 →", A & A == A)
print("Поглинання:       A & set() == set()         →", A & set() == set())
print("Розмір:           len(A & B) <= min(len(A), len(B)) →", len(A & B) <= min(len(A), len(B)))
```

### 8.3. Різниця $A \setminus B$ — «є в першій, немає в другій»

$$A \setminus B = \{x \mid x \in A \ \text{і} \ x \notin B\}$$

З першої множини прибирають усе, що є в другій. Елементи другої множини, яких немає в першій, на результат **не впливають**.

**Запитання, на яке відповідає:** «що залишилося зробити?», «чого бракує?», «що з'явилося нового?».

**Властивості — тут більше «не», ніж «так»:**

| Властивість | Запис | Що означає на практиці |
|---|---|---|
| **Не** комутативна | $A \setminus B \ne B \setminus A$ | порядок множин змінює зміст запитання |
| **Не** асоціативна | $(A \setminus B) \setminus C \ne A \setminus (B \setminus C)$ | ланцюжок `A - B - C` рахується зліва направо |
| Різниця із собою | $A \setminus A = \varnothing$ | |
| Різниця з порожньою | $A \setminus \varnothing = A$ | |
| Зв'язок з об'єднанням | $A \setminus B \setminus C = A \setminus (B \cup C)$ | відняти кілька множин = відняти їх об'єднання |
| Розмір | $\lvert A \setminus B\rvert = \lvert A\rvert - \lvert A \cap B\rvert$ | |

**У Python:** `A - B` або `A.difference(B, C, ...)`.

```python
all_tasks = {"звіт", "тести", "рев'ю", "деплой"}
done_tasks = {"тести", "звіт", "нарада"}          # «нарада» не входила в план

print("Залишилось зробити (план - зроблене):", sorted(all_tasks - done_tasks))
print("Зроблено поза планом (зроблене - план):", sorted(done_tasks - all_tasks))
print()
print("A - A:    ", all_tasks - all_tasks)
print("A - set():", sorted(all_tasks - set()))
```

```python
A, B, C = {1, 2, 3, 4, 5}, {2, 3}, {3, 4}

print("(A - B) - C =", sorted((A - B) - C))
print("A - (B - C) =", sorted(A - (B - C)), " ← інший результат: різниця не асоціативна")
print("A - B - C   =", sorted(A - B - C), " ← Python рахує зліва направо")
print("A - (B | C) =", sorted(A - (B | C)), " ← те саме, що A - B - C")
print()
print("Розмір: len(A - B) == len(A) - len(A & B) →", len(A - B) == len(A) - len(A & B))
```

### 8.4. Симетрична різниця $A \,\triangle\, B$ — «рівно в одній із двох»

$$A \,\triangle\, B = (A \setminus B) \cup (B \setminus A) = (A \cup B) \setminus (A \cap B)$$

Результат містить елементи, які є **в одній** із множин, але **не в обох**. Це «виключне або» (XOR) для множин.

**Запитання, на яке відповідає:** «чим ці дві множини відрізняються?» — наприклад, що змінилося між двома версіями списку.

**Властивості:**

| Властивість | Запис | Що означає на практиці |
|---|---|---|
| Комутативність | $A \,\triangle\, B = B \,\triangle\, A$ | порядок не важливий |
| Асоціативність | $(A \,\triangle\, B) \,\triangle\, C = A \,\triangle\, (B \,\triangle\, C)$ | |
| Із собою | $A \,\triangle\, A = \varnothing$ | однакові множини не мають відмінностей |
| Нейтральний елемент | $A \,\triangle\, \varnothing = A$ | |
| Оборотність | $(A \,\triangle\, B) \,\triangle\, B = A$ | повторне застосування повертає початкову множину |
| Розмір | $\lvert A \,\triangle\, B\rvert = \lvert A\rvert + \lvert B\rvert - 2\lvert A \cap B\rvert$ | |

**Пастка з трьома множинами.** `A ^ B ^ C` — це **не** «елементи, які є рівно в одній із трьох». Це елементи, які трапляються в **непарній** кількості множин: в одній або в усіх трьох.

**У Python:** `A ^ B` або `A.symmetric_difference(B)` — метод приймає лише **один** аргумент.

```python
config_old = {"debug", "cache", "logging"}
config_new = {"cache", "logging", "metrics", "tracing"}

changed = config_old ^ config_new
print("Чим відрізняються версії:", sorted(changed))
print("  прибрали:", sorted(config_old - config_new))
print("  додали:  ", sorted(config_new - config_old))

A, B = config_old, config_new
print()
print("A ^ B == (A - B) | (B - A) →", A ^ B == (A - B) | (B - A))
print("A ^ B == (A | B) - (A & B) →", A ^ B == (A | B) - (A & B))
print("(A ^ B) ^ B == A           →", (A ^ B) ^ B == A)
```

```python
A, B, C = {1, 2, 3}, {2, 3, 4}, {3, 4, 5}

print("A ^ B ^ C =", sorted(A ^ B ^ C))
print("  3 є в усіх трьох множинах — і він у результаті")
print("  2 є у двох множинах (A і B) — його немає")

exactly_one = (A | B | C) - (A & B) - (A & C) - (B & C)
print()
print("Рівно в одній із трьох:", sorted(exactly_one))
```

### 8.5. Відношення між множинами: рівність, підмножина, надмножина

Ці операції нічого не будують — вони відповідають `True` або `False`.

| Математика | Python | Читається |
|---|---|---|
| $A = B$ | `A == B` | ті самі елементи (порядок і спосіб створення не важливі) |
| $A \ne B$ | `A != B` | є хоч один елемент, який належить лише одній |
| $A \subseteq B$ | `A <= B`, `A.issubset(B)` | кожен елемент $A$ є в $B$ (множини можуть бути рівні) |
| $A \subset B$ | `A < B` | $A \subseteq B$ **і** $A \ne B$ — **власна** підмножина |
| $A \supseteq B$ | `A >= B`, `A.issuperset(B)` | $A$ містить усі елементи $B$ |
| $A \supset B$ | `A > B` | $A \supseteq B$ **і** $A \ne B$ — **власна** надмножина |
| $A \cap B = \varnothing$ | `A.isdisjoint(B)` | немає жодного спільного елемента |

**Три факти, які варто пам'ятати:**

1. Порожня множина — підмножина будь-якої: $\varnothing \subseteq A$.
2. Кожна множина — підмножина самої себе: $A \subseteq A$, але $A \subset A$ хибне.
3. Множини впорядковані лише **частково**. Для чисел завжди істинне одне з трьох: `a < b`, `a == b`, `a > b`. Для множин усі три можуть бути `False` одночасно — якщо жодна не міститься в іншій. Тому сортувати список множин через `sorted()` не має сенсу.

```python
required = {"id", "name"}
provided = {"id", "name", "email"}

print("required <= provided  (підмножина):        ", required <= provided)
print("required <  provided  (власна підмножина): ", required < provided)
print("provided >= required  (надмножина):        ", provided >= required)
print("required <= required  (сама собі):         ", required <= required)
print("required <  required  (власна сама собі):  ", required < required)
print("set() <= required     (порожня множина):   ", set() <= required)
print()
print("{1, 2, 3} == {3, 2, 1, 1} →", {1, 2, 3} == {3, 2, 1, 1})
```

```python
A, B = {1, 2}, {2, 3}             # перетинаються, але жодна не міститься в іншій

print("A <  B →", A < B)
print("A == B →", A == B)
print("A >  B →", A > B)
print("Усі три False: ці множини не можна порівняти «більше-менше».")
print()

# Як підмножина пов'язана з іншими операціями
S, T = {1, 2}, {1, 2, 3}
print("S <= T                →", S <= T)
print("S | T == T  (те саме) →", S | T == T)
print("S & T == S  (те саме) →", S & T == S)
print("S - T == set()        →", S - T == set())
```

### 8.6. Належність $x \in A$ і потужність $\lvert A \rvert$

- `x in A` / `x not in A` — чи належить елемент множині. Завдяки хеш-таблиці час перевірки не залежить від розміру множини.
- `len(A)` — **потужність**: кількість елементів.

Потужності пов'язує **формула включень-виключень**. Якщо просто додати розміри двох множин, спільні елементи порахуються двічі — тому їх треба один раз відняти:

$$\lvert A \cup B\rvert = \lvert A\rvert + \lvert B\rvert - \lvert A \cap B\rvert$$

Для трьох множин:

$$\lvert A \cup B \cup C\rvert = \lvert A\rvert + \lvert B\rvert + \lvert C\rvert - \lvert A \cap B\rvert - \lvert A \cap C\rvert - \lvert B \cap C\rvert + \lvert A \cap B \cap C\rvert$$

На практиці це відповідь на запитання «скільки **різних** людей у двох списках», коли відомі лише розміри списків і кількість збігів.

```python
python_course = {"Оля", "Марко", "Іра", "Денис"}
sql_course = {"Іра", "Денис", "Софія"}
git_course = {"Софія", "Тарас", "Іра"}
A, B, C = python_course, sql_course, git_course

print("'Іра' in A:        ", "Іра" in A)
print("'Тарас' not in A:  ", "Тарас" not in A)
print()
print(f"|A| = {len(A)}, |B| = {len(B)}, |A ∩ B| = {len(A & B)}")
print(f"|A ∪ B| напряму:    {len(A | B)}")
print(f"|A| + |B| - |A ∩ B|: {len(A) + len(B) - len(A & B)}")

by_formula = (len(A) + len(B) + len(C)
              - len(A & B) - len(A & C) - len(B & C)
              + len(A & B & C))
print()
print(f"Три множини: напряму {len(A | B | C)}, за формулою {by_formula}")
```

### 8.7. Доповнення $\overline{A}$ і закони де Моргана

**Доповнення** — це «все, що не входить в $A$». Фраза має зміст лише тоді, коли відомо, що таке «все», — тобто задано універсум $U$:

$$\overline{A} = U \setminus A$$

У Python окремого оператора доповнення **немає**, бо множина не знає свого універсуму. Доповнення завжди пишуть як різницю: `U - A`.

**Властивості:** $\overline{\overline{A}} = A$, $\quad A \cup \overline{A} = U$, $\quad A \cap \overline{A} = \varnothing$.

**Закони де Моргана** пов'язують доповнення з об'єднанням і перетином:

$$\overline{A \cup B} = \overline{A} \cap \overline{B} \qquad\qquad \overline{A \cap B} = \overline{A} \cup \overline{B}$$

Словами: «не (A або B)» — це «не A **і** не B»; «не (A і B)» — це «не A **або** не B». Та сама логіка працює для `not (a or b)` і `not (a and b)` в умовах `if`.

```python
U = set(range(1, 11))                 # універсум: числа від 1 до 10
even = {2, 4, 6, 8, 10}
small = {1, 2, 3, 4}

not_even = U - even
print("Доповнення парних:", sorted(not_even))
print("U - (U - A) == A      →", U - (U - even) == even)
print("A | (U - A) == U      →", even | (U - even) == U)
print("A & (U - A) == set()  →", even & (U - even) == set())
```

```python
A, B = even, small

left_1, right_1 = U - (A | B), (U - A) & (U - B)
left_2, right_2 = U - (A & B), (U - A) | (U - B)

print("не (A або B):       ", sorted(left_1))
print("(не A) і (не B):    ", sorted(right_1), " → рівні:", left_1 == right_1)
print()
print("не (A і B):         ", sorted(left_2))
print("(не A) або (не B):  ", sorted(right_2), " → рівні:", left_2 == right_2)
```

### 8.8. Закони, що пов'язують операції між собою

Об'єднання й перетин поводяться подібно до додавання й множення чисел — але з власними законами.

| Закон | Запис |
|---|---|
| Дистрибутивність перетину | $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$ |
| Дистрибутивність об'єднання | $A \cup (B \cap C) = (A \cup B) \cap (A \cup C)$ |
| Поглинання | $A \cup (A \cap B) = A$, $\quad A \cap (A \cup B) = A$ |
| Різниця через перетин | $A \setminus B = A \cap \overline{B}$ |
| Розклад об'єднання на частини | $A \cup B = (A \setminus B) \cup (A \cap B) \cup (B \setminus A)$ — три частини без спільних елементів |

Другий закон дистрибутивності у чисел **не діє** ($a + b \cdot c \ne (a + b)\cdot(a + c)$), а в множин — діє. Це одна з причин не переносити інтуїцію арифметики на множини наосліп.

**Навіщо це в коді.** Закони дозволяють переписати вираз коротше або дешевше. Наприклад, замість двох перетинів і об'єднання `(A & B) | (A & C)` досить одного перетину `A & (B | C)`.

```python
U = set(range(1, 11))
A, B, C = {1, 2, 3, 4, 5}, {4, 5, 6, 7}, {1, 5, 7, 9}

print("A & (B | C) == (A & B) | (A & C) →", A & (B | C) == (A & B) | (A & C))
print("A | (B & C) == (A | B) & (A | C) →", A | (B & C) == (A | B) & (A | C))
print("A | (A & B) == A                 →", A | (A & B) == A)
print("A & (A | B) == A                 →", A & (A | B) == A)
print("A - B == A & (U - B)             →", A - B == A & (U - B))

only_a, both, only_b = A - B, A & B, B - A
print()
print("лише A:", sorted(only_a), "| спільні:", sorted(both), "| лише B:", sorted(only_b))
print("разом це A | B →", only_a | both | only_b == A | B,
      "; частини не перетинаються →", only_a.isdisjoint(both) and both.isdisjoint(only_b) and only_a.isdisjoint(only_b))
```

### 8.9. Операції без власного оператора: декартів добуток і булеан

**Декартів добуток** — множина всіх **упорядкованих пар**, де перший елемент з $A$, другий з $B$:

$$A \times B = \{(a, b) \mid a \in A,\ b \in B\} \qquad \lvert A \times B\rvert = \lvert A\rvert \cdot \lvert B\rvert$$

Він не комутативний: пара `("S", "чорний")` — не те саме, що `("чорний", "S")`. У Python — `itertools.product(A, B)` або вкладений генератор.
Де трапляється: усі комбінації розміру й кольору, усі клітинки дошки, усі пари «користувач — товар».

**Булеан** (*power set*) — множина **всіх підмножин**, разом із порожньою і самою $A$:

$$\mathcal{P}(A) = \{S \mid S \subseteq A\} \qquad \lvert \mathcal{P}(A)\rvert = 2^{\lvert A\rvert}$$

Готової функції немає; булеан збирають з `itertools.combinations` для кожного розміру від 0 до $\lvert A\rvert$. Підмножини зберігають як `frozenset`, бо звичайну множину не можна покласти всередину іншої множини.
Розмір росте дуже швидко: для 20 елементів це вже понад мільйон підмножин.

```python
from itertools import product

sizes = {"S", "M"}
colors = {"чорний", "білий", "синій"}

variants = set(product(sizes, colors))
print("Усі варіанти футболки (розмір, колір):")
for size, color in sorted(variants):
    print(f"  {size} / {color}")

print()
print(f"|A × B| = {len(variants)} = {len(sizes)} · {len(colors)}")
print("A × B == B × A →", variants == set(product(colors, sizes)), " (порядок у парі має значення)")
print("Те саме генератором:", variants == {(s, c) for s in sizes for c in colors})
```

```python
from itertools import combinations


def power_set(items):
    # Усі підмножини: по одній групі для кожного розміру від 0 до len(items)
    items = sorted(items)
    result = set()
    for size in range(len(items) + 1):
        for combo in combinations(items, size):
            result.add(frozenset(combo))
    return result


toppings = {"сир", "гриби", "шинка"}
subsets = power_set(toppings)

for subset in sorted(subsets, key=lambda s: (len(s), sorted(s))):
    print(sorted(subset) if subset else "∅ (без добавок)")

print()
print(f"|P(A)| = {len(subsets)} = 2 ** {len(toppings)}")
print("Для 20 елементів підмножин уже:", 2 ** 20)
```

### 8.10. Оператор, метод чи версія «на місці»: що обрати

Одну й ту саму математичну операцію Python дає у трьох формах. Вони відрізняються не результатом, а поведінкою.

| Форма | Приклад | Приймає | Що відбувається з лівою множиною |
|---|---|---|---|
| Оператор | `A \| B` | **тільки** `set` / `frozenset` з обох боків | не змінюється, створюється нова |
| Метод | `A.union(B, C)` | будь-який ітерований об'єкт: список, кортеж, рядок, генератор | не змінюється, створюється нова |
| Оператор «на місці» | `A \|= B` | тільки `set` / `frozenset` | **змінюється сама** |
| Метод «на місці» | `A.update(B, C)` | будь-який ітерований об'єкт | **змінюється сама**, метод повертає `None` |

**Чотири пастки:**

1. `A | [1, 2]` → `TypeError`. Оператор не перетворює список на множину; метод `A.union([1, 2])` — перетворює.
2. `A = A.update(B)` → в `A` опиниться `None`. Методи «на місці» нічого не повертають.
3. `B = A`, а потім `A |= C` — зміниться і `B`, бо це той самий об'єкт. А `A = A | C` створює нову множину, і `B` лишається старою.
4. `A.union("кава")` додасть **окремі літери**, бо рядок — це послідовність символів. Один рядок передають як `A.union(["кава"])` або `A | {"кава"}`.

**`frozenset`** підтримує всі операції, що створюють нову множину, але не має жодної версії «на місці». Якщо змішати `set` і `frozenset`, тип результату визначає **лівий** операнд.

**Скільки це коштує** (середній випадок; `len(A)` = $n$, `len(B)` = $m$):

| Операція | Час |
|---|---|
| `x in A` | $O(1)$ |
| `A \| B` | $O(n + m)$ |
| `A & B` | $O(\min(n, m))$ |
| `A - B` | $O(n)$ |
| `A ^ B` | $O(n + m)$ |
| `A <= B` | $O(n)$ |
| `A.isdisjoint(B)` | $O(\min(n, m))$, зупиняється на першому збігу |

```python
A = {1, 2, 3}

try:
    A | [3, 4]
except TypeError as error:
    print("A | [3, 4]        → TypeError:", error)

print("A.union([3, 4])   →", sorted(A.union([3, 4])))
print("A.union((4,), range(5, 7), {7}) →", sorted(A.union((4,), range(5, 7), {7})))
menu = {"чай"}
print()
print('menu.union("кава")   →', sorted(menu.union("кава")), " ← окремі літери!")
print('menu.union(["кава"]) →', sorted(menu.union(["кава"])))
print('menu | {"кава"}      →', sorted(menu | {"кава"}))
print()
print("Результат update():", A.update([9]), " ← None, змінилася сама множина:", sorted(A))
```

```python
A = {1, 2}
B = A                       # друге ім'я для ТІЄЇ САМОЇ множини

A |= {3}                    # змінює об'єкт на місці
print("після A |= {3}:    A =", sorted(A), " B =", sorted(B), " той самий об'єкт:", A is B)

A = A | {4}                 # створює НОВУ множину і вішає на неї ім'я A
print("після A = A | {4}: A =", sorted(A), " B =", sorted(B), " той самий об'єкт:", A is B)

frozen = frozenset({1, 2})
print()
print("frozenset | set →", type(frozen | {3}).__name__)
print("set | frozenset →", type({3} | frozen).__name__)
print("frozenset має update:", hasattr(frozen, "update"))
```

### 8.11. Підсумок: від запитання до операції

Операцію обирають не за назвою, а за запитанням, на яке треба відповісти.

| Запитання до даних | Операція | Код |
|---|---|---|
| Хто є хоча б в одному списку? | об'єднання | `A \| B` |
| Хто є в обох? | перетин | `A & B` |
| Хто є в першому, але не в другому? | різниця | `A - B` |
| Чим два списки відрізняються? | симетрична різниця | `A ^ B` |
| Хто **не** входить у групу? | доповнення | `U - A` |
| Чи всі потрібні присутні? | підмножина | `required <= provided` |
| Чи є хоч щось спільне? | неперетинність | `not A.isdisjoint(B)` |
| Скільки різних? | потужність | `len(A \| B)` |
| Усі можливі пари? | декартів добуток | `product(A, B)` |

Остання комірка збирає кілька операцій в одну практичну задачу: порівняння двох знімків бази користувачів.

```python
users_monday = {"anna", "bohdan", "ira", "maks", "olha"}
users_friday = {"bohdan", "ira", "olha", "sofia", "taras"}
admins = {"olha", "root"}

print("Були й залишились:      ", sorted(users_monday & users_friday))
print("Пішли за тиждень:       ", sorted(users_monday - users_friday))
print("Нові за тиждень:        ", sorted(users_friday - users_monday))
print("Усього змін:            ", len(users_monday ^ users_friday))
print("Усі, хто був за тиждень:", sorted(users_monday | users_friday))
print()
print("Усі адміни є серед користувачів п'ятниці:", admins <= users_friday)
print("Адміни, яких немає в базі:               ", sorted(admins - users_friday))
print("Серед новеньких є адміни:                ", not admins.isdisjoint(users_friday - users_monday))
```
